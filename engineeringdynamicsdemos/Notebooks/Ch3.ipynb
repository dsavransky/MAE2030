## Simple Pendulum Animation

See Example 3.11. Execute the following cell to view the animation. You can change the resolution of the generated figure by changing the value in the first line of code, and the time array `t` (duration and step) of the integration; longer/finer arrays take longer to render.

```python
plt.rcParams["figure.dpi"] = 50
anim = ch3.simple_pendulum_animation(t=np.arange(0, 10, 1 / 20))
plt.close(anim._fig)
HTML(anim.to_jshtml())
```

## Stationary vs. Accelerating Pendulum

See Example 3.14. A stationary pendulum beside one in an accelerating box. With `stationary=True` the accelerating pendulum starts at rest in the accelerating frame. Execute the following cell to view the animation. You can change the resolution of the generated figure by changing the time array `t` (duration and step) of the integration; longer/finer arrays take longer to render.

```python
anim = ch3.accel_pendulum_animation(t=np.arange(0, 5, 1 / 20))
plt.close(anim._fig)
HTML(anim.to_jshtml())
```

```python
anim = ch3.accel_pendulum_animation(stationary=True, t=np.arange(0, 5, 1 / 20))
plt.close(anim._fig)
HTML(anim.to_jshtml())
```

## Spring Pendulum Animation

Animates a pendulum whose arm is a spring, alongside live traces of the arm length `r` and angle `theta`. Execute the following cell to view the animation. ## Rotating Frame Animation

See Sec. 3.4.2. A rigid body (cube) fixed in a frame $\mathcal{B}$ rotating about $\mathbf{\hat{e}}_3$, as seen in the inertial frame $\mathcal{I}$. `step_deg` sets the angular increment per frame (smaller is smoother but slower to render).

```python
anim = ch3.rotating_frame_animation(step_deg=6)
plt.close(anim._fig)
HTML(anim.to_jshtml())
```
