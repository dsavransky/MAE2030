## Mass-Spring SHO Animation

See Tutorial 2.3. Animates a mass-spring simple harmonic oscillator: a schematic view of the block/spring alongside live position and velocity traces.

Execute the following cell to view the animation.  You can change the resolution of the generated figure by changing the value in the first line of code, the maximum time of the integration by setting the `t_max` parameter, and the frame rate by changing the `dt` parameter.  Note that changing any of these can significantly increase the amount of time it takes to generate the animation.

```python
plt.rcParams["figure.dpi"] = 50
anim = ch2.mass_spring_sho_animation(t_max=3.0, dt=1 / 23)
plt.close(anim._fig)
HTML(anim.to_jshtml())
```
