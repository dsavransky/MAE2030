## Mass-Spring SHO Animation

See Tutorial 2.3. Animates a mass-spring simple harmonic oscillator: a schematic view of the block/spring alongside live position and velocity traces.

`to_jshtml` embeds every animation frame as an individual image in the notebook, so `t_max` is shortened from the ported default of 60s to 3s here (still ~5 full oscillation cycles at `w0=10` rad/s), the figure DPI is lowered from the default, and `dt` is set to target an effective ~23 fps playback rate (vs. the ported default of 40 fps) - purely to keep `to_jshtml`'s generation time and the embedded animation's output size reasonable for committing to git. None of these changes affect the ported `mass_spring_sho_animation` function itself.

```python
plt.rcParams["figure.dpi"] = 50
anim = ch2.mass_spring_sho_animation(t_max=3.0, dt=1 / 23)
plt.close(anim._fig)
HTML(anim.to_jshtml())
```
